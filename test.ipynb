```python
import pysrt

def get_text(srt_file_path):
    subtitles = pysrt.open(srt_file_path)
    all_text=""
    for sub in subtitles:
        all_text+=f"{sub.index} - {sub.start} --> {sub.end}\n{sub.text}\n"
    return  all_text
    

print(get_text("srt_generation/generated_srt.srt"))
    
```

```python
import pysrt
from chains import generate_topic
def read_srt(file_path):
    subs = pysrt.open(file_path)
    # for sub in subs:
    #     print(f"{sub.index} - {sub.start} --> {sub.end}\n{sub.text}\n")
    return subs

def subtitle_data(file_name):
    subtitle_data=[]  

    for sub in read_srt(file_name):

        subtitle_data.append({
            "start": sub.start.to_time().strftime("%H:%M:%S.%f")[:-3],
            "end": sub.end.to_time().strftime("%H:%M:%S.%f")[:-3],
            "text": sub.text.replace("\n", " ")  
        })
    return subtitle_data
```
